```python
import pandas as pd
import csv
import os
from datetime import datetime
import random
import nltk
from nltk.corpus import names

nltk.download('names')
```

```python
# Label each name with its gender
labeled_names = ([(name, 'male') for name in names.words('male.txt')] +
                 [(name, 'female') for name in names.words('female.txt')])

# Shuffle with a fixed seed so your splits are reproducible
random.seed(42)
random.shuffle(labeled_names)
```

```python
RESULTS_FILE = "project3_results.csv"
results = []   # in-memory log of every experiment
```

```python
# Split: 500 test, 500 dev-test, remainder train
test_names    = labeled_names[:500]
devtest_names = labeled_names[500:1000]
train_names   = labeled_names[1000:]

print(f"Total:    {len(labeled_names)}")
print(f"Test:     {len(test_names)}")
print(f"Dev-test: {len(devtest_names)}")
print(f"Train:    {len(train_names)}")
```

```python
# Baseline feature extractor (from the chapter)
def gender_features(name):
    return {'last_2_letters': name[-2]}
```

```python
# Convert each (name, gender) pair into a (features, gender) pair
train_set   = [(gender_features(n), g) for (n, g) in train_names]
devtest_set = [(gender_features(n), g) for (n, g) in devtest_names]
```

```python
train_set
```

```python
# Train the Naive Bayes classifier
classifier = nltk.NaiveBayesClassifier.train(train_set)
```

```python
# Evaluate on the dev-test set (NOT the test set)
print("Dev-test accuracy:", nltk.classify.accuracy(classifier, devtest_set))
```

```python
# Look at the most informative features
classifier.show_most_informative_features(10)
```

```python
for name in ['Neo', 'Trinity', 'Morpheus', 'Alexandra']:
    print(name, '->', classifier.classify(gender_features(name)))
```

```python
errors = []
for (name, tag) in devtest_names:
    guess = classifier.classify(gender_features(name))
    if guess != tag:
        errors.append((tag, guess, name))

for (tag, guess, name) in sorted(errors)[:40]:
    print(f'correct={tag:<8} guess={guess:<8} name={name}')
```
