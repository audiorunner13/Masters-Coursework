Frequency Distributions

```python
# Tells the frequency of each vocabulary item in the test
fdist1 = FreqDist(text1)
fdist1
```

```python
# keys() gives a list of all the distinct types in the text (words, punctuations, etc.)
vocabulary1 = fdist1.keys()
vocabulary1
```

```python
# Give the first 50 distinct types int he text by slicing the list
list(vocabulary1)[:50]
```

```python
# Gives the number of time the work whale appears in text1 - Moby Dick
fdist1['whale']
```

```python
#Generate a cumulative frequency plot for all words in text1
fdist1.plot(50,cumulative=True)
```

Fine-Grained Selection of Words

```python
# find the words from the vocabulary of the text that are more than 15 Charactors long.
V=set(text1)
long_words = [w for w in V if len(w) > 15]
sorted(long_words)
```

```python
# find the words from the vocabulary of the text that are more than 10 Charactors long.
V=set(text1)
long_words = [word for word in V if len(word) > 10]
sorted(long_words)
```

```python
sorted([w for w in set(text5) if len(w) > 7 and fdist5[w] > 7])
```
